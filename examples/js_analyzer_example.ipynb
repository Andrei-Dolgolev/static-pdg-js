## Import Required Modules

```python
import os
import sys

# Add the repository root to the path if needed
if '..' not in sys.path:
    sys.path.insert(0, '..')

from src.build_pdg import get_data_flow
from src.display_graph import draw_ast, draw_cfg, draw_pdg
```

## Use Existing Sample JavaScript File

We'll use the existing sample.js file in the examples directory:

```python
# Path to the sample JavaScript file
sample_file = 'sample.js'

# Display the contents of the file
with open(sample_file, 'r') as f:
    js_code = f.read()
    
print(js_code)
```

## Generate and Visualize the PDG

Now we'll use the static-pdg-js package to analyze the JavaScript code:

```python
# Define an output directory for visualizations
output_dir = 'notebook_output'
os.makedirs(output_dir, exist_ok=True)

# Analyze the sample file
print(f"Analyzing {sample_file}...")
pdg = get_data_flow(sample_file, benchmarks=dict())

if pdg:
    # Generate AST visualization
    ast_output = os.path.join(output_dir, 'sample_ast')
    print(f"Generating AST visualization: {ast_output}.pdf")
    draw_ast(pdg, attributes=True, save_path=ast_output)
    
    # Generate CFG visualization
    cfg_output = os.path.join(output_dir, 'sample_cfg')
    print(f"Generating CFG visualization: {cfg_output}.pdf")
    draw_cfg(pdg, attributes=True, save_path=cfg_output)
    
    # Generate PDG visualization
    pdg_output = os.path.join(output_dir, 'sample_pdg')
    print(f"Generating PDG visualization: {pdg_output}.pdf")
    draw_pdg(pdg, attributes=True, save_path=pdg_output)
    
    print("Visualization complete!")
else:
    print("Error: Failed to build PDG")
```

## Display the Visualizations in the Notebook

We can display the generated PDFs directly in the notebook:

```python
from src.display_graph import produce_ast, produce_cfg, produce_pdg

# Generate SVGs directly
def create_svg_visualization(pdg, output_dir):
    # Create SVG for AST
    dot_ast = produce_ast(pdg, attributes=True)
    dot_ast.format = 'svg'
    svg_ast_path = f"{output_dir}/sample_ast.svg"
    dot_ast.render(svg_ast_path, cleanup=True)
    
    # Create SVG for CFG
    dot_cfg = produce_cfg(pdg, attributes=True)
    dot_cfg.format = 'svg'
    svg_cfg_path = f"{output_dir}/sample_cfg.svg"
    dot_cfg.render(svg_cfg_path, cleanup=True)
    
    # Create SVG for PDG
    dot_pdg = produce_pdg(pdg, attributes=True)
    dot_pdg.format = 'svg'
    svg_pdg_path = f"{output_dir}/sample_pdg.svg"
    dot_pdg.render(svg_pdg_path, cleanup=True)
    
    return svg_ast_path, svg_cfg_path, svg_pdg_path

# Create SVGs
svg_ast, svg_cfg, svg_pdg = create_svg_visualization(pdg, output_dir)

# Display SVGs directly in the notebook
from IPython.display import SVG, display, HTML, Markdown

display(Markdown("## AST Visualization"))
display(SVG(f"{svg_ast}.svg"))

display(Markdown("## CFG Visualization"))
display(SVG(f"{svg_cfg}.svg"))

display(Markdown("## PDG Visualization"))
display(SVG(f"{svg_pdg}.svg"))
```
